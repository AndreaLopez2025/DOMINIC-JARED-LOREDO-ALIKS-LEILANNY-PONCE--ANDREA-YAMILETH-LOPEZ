### **4. BUCLES**

```python
############################# BUCLES FOR IN #####################################
#Sintaxis:
#For elemento in elementos:
#   Acciones

frutas=["manzana","pera","uva"]
for fruta in frutas:
    print(fruta)

############################# BUCLES WHILE #####################################
#Sintaxis:
#While condicion:
#   Acciones
contador= 0
while contador<5:
    print(f"contador:{contador}") # Corrected: used f-string and closed string literal
    contador+=1


############################# SENTENCIAS BREAK, PASS Y CONTINUE #####################################

#Break: Detiene la ejecución del bucle.
for numero in range(10):
  if numero==5: # Corrected: added space
    break # Corrected: proper indentation for break
  print(numero) # Moved print outside the if, inside the loop, to print numbers before break

#Pass: Se utiliza para dejar vacío un trozo de código que todavía no queremos especificar.
for i in range(3):
  pass

#Continue: Pasa a la siguiente iteración del bucle.
for numero in range(5): # Corrected: added colon
  if numero==2:
    continue # Corrected: proper indentation for continue
  print(numero)
```

### **6. PROGRAMACIÓN ORIENTADA A OBJETOS**

```python

# Sintaxis:

# class NombreClase():
#   Atributos
#   def __init__(self):
#       Inicializar atributos (constructor)
#   Setters and getters
#   Metodos de clase
#

# Implementación de la clase mascota para un videojuego de animales.

class Mascota:
    ############################# ATRIBUTOS DE LA CLASE #####################################
    # En Python, los atributos de instancia se definen dentro del constructor __init__.

    ############################# CONSTRUCTOR DE LA CLASE #####################################
    def __init__(self, nombre: str, edad: int, salud: int):
        self.__nombre = nombre # Atributo privado
        self.__edad = edad     # Atributo privado
        self.__salud = salud   # Atributo privado

    ############################# METODOS GETTERS AND SETTERS #####################################

    # Los metodos getters y setters sirven para ofrecer la funcionalidad de obtener el valor o cambiar el valor de los atributos que queremos darle
    # la posibilidad al programador que utilice la clase.

    # Sintaxis getters: (Los atributos deben ser privados o protegidos, sino no tendría sentido hacer un get o set ya que podemos modificar y
    # consultar los atributos directamente, es más daría un bucle infinito.)

    # def get_atributo(self) -> tipo_atributo:
    #   return self.__nombre_atributo

    # Sintaxis setters:
    # def set_atributo(self, value) -> None:
    #   self.__nombre_atributo = value

    # Getter para nombre
    def get_nombre(self) -> str:
        return self.__nombre

    # Getter para edad
    def get_edad(self) -> int:
        return self.__edad

    # Setter para salud
    def set_salud(self, salud: int) -> None:
        self.__salud = salud

    ############################# METODOS DE CLASE #####################################

    # Método que sirve para aumentar la salud a tu mascota.
    def alimentar(self) -> None:
        self.__salud += 10
        print(f"La mascota fue alimentada. Salud actual: {self.__salud}")

    # Método que sirve para aumentar de nivel a tu mascota en función del torneo que haya ganado.
    def ganar_torneo(self) -> None:
        self.__edad += 1
        print(f"La mascota ganó un torneo. Edad actual: {self.__edad}")

# Ejemplo de uso de la clase Mascota:
# mi_mascota = Mascota("Fido", 3, 100)
# print(mi_mascota.get_nombre())
# mi_mascota.alimentar()
# mi_mascota.ganar_torneo()
# print(mi_mascota.get_edad())
```
